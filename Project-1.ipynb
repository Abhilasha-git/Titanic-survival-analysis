<font size="7"><b>DROP CABIN COLUMN</b></font>


```python
if 'Cabin' in df.columns:
    df.drop('Cabin', axis=1, inplace=True)
    print("Cabin column dropped successfully.")
else:
    print("Cabin column not found (it might already be dropped).")

print(df.head())
```

<font size="7"><b>CHANGES</b></font>


```python
# Calculate the median age and fill the missing values
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print(df.head())
```

<font size="7"><b>SURVIVAL RATE BY AGE</b></font>

```python
# Example for the Age distribution plot we made earlier:

sns.kdeplot(df['Age'][df['Survived'] == 0], label='Did Not Survive', color='Red', fill=True)
sns.kdeplot(df['Age'][df['Survived'] == 1], label='Survived', color='Blue', fill=True)


plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Density')

# Add the legend here:
plt.legend() 

plt.show()
```

<font size="7"><b>SURVIVAL RATE BY GENDER</b></font>

```python
plt.figure(figsize=(4, 3))
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival Rate by Gender')
plt.xlabel('Gender (Male=0, Female=1)')
plt.ylabel('Survival Rate')
plt.show()
```

<font size="7"><b> SURVIVAL RATE BY PASSENGER CLASS</b></font>

```python
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title('Survival Rate by Pclass')
plt.show()
```
